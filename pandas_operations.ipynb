## Pandas Operations Practice



```python
import pandas as pd 
```

### Reading CSV Files


```python
df=pd.read_csv("data.csv")
print(df.head())
```

### Selecting Rows and Columns


```python

#Select a column
print(df["Date"]) 

#Select multiple columns
print(df[["Date","Sales"]]) 
```

### Filtering Data


```python
filter_df=df[df["Category"]=="A"]
print(filter_df)
```

### Handling Missing Values

```python
#Check missing values
print(df.isnull().sum()) 
```

```python

#Fill missing values with mean of that column

df["Sales_fill"]=df["Sales"].fillna(df["Sales"].mean()) 
df
df["Value_fill"]=df["Value"].fillna(df["Value"].mean()) 
df
```

### Sorting Data

```python
df_sort=df.sort_values(by="Category")
print(df_sort)
```

### Basic Data Analysis


```python
print("Shape:", df.shape)

print("Columns:")
print(df.columns)

print("Summary Statistics:")
print(df.describe())

print("Maximum Sales:")
print(df["Sales"].max())

print("Average Sales:")
print(df["Sales"].mean())
```
